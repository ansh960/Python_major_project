```python
x=30 # Global Variable
def show():
    x=20 # Local Variable
    print("Local Variable: ",x)# 20
show()

print("Global Variable: ",x) #30

    
```

```python
x=30 #Global Variable
def show():
    global x
    print("global Variable:",x) #20
show()
print("Global Variable: ",x) #30

def display():
    global x
    print("Global Variable: ",x) #30
```

```python
balance=3000 #Global Variable
def show ():
    global balance 
    deposit=1000
    print("Deposit Balance: ",deposit)

    balance=balance+deposit #4000

show()
print("Current Blance After Deposit: ",balance)
```

```python
balance=3000 #global Variable
def show(): #function without parameter
    deposit=1000
    print("Deposit Balance: ", deposit)
    print("Initial Balance Was: ",globals() ["balance"]) #one time use

show()
print("Current Balance After Deposit: ",balance)
```

```python
def add(*args):
    bill=0
    for price in args:
        bill=bill+price
    print("Bill: ",bill)
add(300,300,300)
```

```python
#args
def add(*args): #tuple
    total=0 #30
    for ex in args:
        total=total+ex  #150
    print("Sum: ",total)

add(10,20)
add(40,50,60)
add(20,30,50,60)
add(20,30,40,50,60,70)

    
```

```python
def add(*args):
    print("agrs[0]",args[0])
    print("agrs[1]",args[1])
    print("agrs[2]",args[2])
    print("agrs[3]",args[3])
    print("agrs[4]",args[4])
    print(args)

add(20,30,40,50,60)

    
```

```python
# *kwargs #keyword Argument >> Dictionary

def Employeedetails(**kwargs):
    print(kwargs)

Employeedetails(name="Mohit",Role="Developer",Department="IT")
Employeedetails(name="Mohit",Department="IT")
Employeedetails(name="Mohit")

```

```python
def Employeedetails(**kwargs):
    print(kwargs['name'])
    print(kwargs['Role'])
    print(kwargs['Department'])
Employeedetails(name="Mohit",Role="Developer",Department="IT")

    
```

```python
def Employeedetails(**kwargs):
    for key, value in kwargs.items():
        print(key, ":",value)

    print("======================")


Employeedetails(name="Mohit",Role="Developer",Department="IT")
Employeedetails(name="Mohit",Department="IT")
Employeedetails(name="Mohit")

```
